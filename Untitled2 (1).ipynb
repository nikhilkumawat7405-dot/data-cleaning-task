```python
import zipfile
import requests
from io import BytesIO
import glob

url = "https://archive.ics.uci.edu/static/public/320/student+performance.zip"

data = requests.get(url).content

with zipfile.ZipFile(BytesIO(data)) as z:
    with zipfile.ZipFile(BytesIO(z.read("student.zip"))) as student_zip:
        student_zip.extractall("student_data")

print(glob.glob("student_data/*"))
```

```python
import pandas as pd

df = pd.read_csv("student_data/student-mat.csv", sep=";")

print("Rows and Columns:", df.shape)
df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Duplicate rows remove karna
df = df.drop_duplicates()

# Missing values remove karna
df = df.dropna()

print("Cleaned dataset shape:", df.shape)
```

```python
df.to_csv("cleaned_student_data.csv", index=False)

print("File saved successfully!")
```
